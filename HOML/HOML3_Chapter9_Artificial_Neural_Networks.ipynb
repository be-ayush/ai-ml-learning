```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris(as_frame=True)
```

```python
X = iris.data[["petal length (cm)", "petal width (cm)"]].values
Y = (iris.target == 0)

perceptron_classifier = Perceptron(random_state=42)
perceptron_classifier.fit(X, Y)

print(perceptron_classifier.score(X, Y))

X_new = [[2, 0.5], [3, 1]]
y_pred = perceptron_classifier.predict(X_new)
y_pred
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data, housing.target, random_state=42
)

mlp_regressor = MLPRegressor(hidden_layer_sizes=[50, 50, 50], early_stopping=True, verbose=True, random_state=42)

pipeline = make_pipeline(StandardScaler(), mlp_regressor)
pipeline.fit(X_train, y_train)
```

```python
mlp_regressor.best_validation_score_
```

```python
y_pred = pipeline.predict(X_test)
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```

```python
from sklearn.datasets import fetch_openml

fashion_mnist = fetch_openml(name="Fashion-MNIST", as_frame = False)
targets = fashion_mnist.target.astype(int)
```

```python
X_train, Y_train = fashion_mnist.data[:60000], targets[:60000]
X_test, Y_test = fashion_mnist.data[60000:], targets[60000:]
```

```python
import plotly.express as px

# Select the first sample from X_train and reshape it to 28x28
sample_image_plotly = X_train[0].reshape(28, 28)

fig = px.imshow(sample_image_plotly, color_continuous_scale='gray')
fig.update_layout(coloraxis_showscale=False) # Hide the color scale
fig.show()
```

```python
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
class_names[Y_train[0]]
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler

mlf_classifier = MLPClassifier(hidden_layer_sizes=[300, 100], verbose=True, early_stopping=True, random_state=42)

pipeline = make_pipeline(MinMaxScaler(), mlf_classifier)
pipeline.fit(X_train, Y_train)
```

```python
accuracy = pipeline.score(X_test, Y_test)
print(f"Accuracy: {accuracy}")
```

```python
X_new = X_test[:15]
y_pred = pipeline.predict(X_new)
y_pred
```

```python
y_proba = mlf_classifier.predict_proba(X_new)
y_proba[12]
```
