```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples = 1000, random_state = 42)

k = 5
kmeans = KMeans(n_clusters=k, random_state=42)
y_pred = kmeans.fit_predict(X)

```

```python
import plotly.express as px
import pandas as pd

# Create a DataFrame for easier plotting with Plotly
df_clusters = pd.DataFrame(X, columns=['Feature 1', 'Feature 2'])
df_clusters['Cluster'] = y_pred.astype(str)

fig = px.scatter(df_clusters, x='Feature 1', y='Feature 2', color='Cluster',
                 title='K-Means Clustering Results (Plotly)',
                 labels={'Feature 1': 'Feature 1', 'Feature 2': 'Feature 2'})

# Add cluster centers to the plot
cluster_centers_df = pd.DataFrame(kmeans.cluster_centers_, columns=['Feature 1', 'Feature 2'])
fig.add_scatter(x=cluster_centers_df['Feature 1'], y=cluster_centers_df['Feature 2'],
                mode='markers', marker=dict(symbol='x', size=15, color='red', line=dict(width=2, color='black')),
                name='Cluster Centers')

fig.update_layout(showlegend=True)
fig.show()
```

```python
import numpy as np
good_initialization = np.array([[-3,3], [-3, 2], [-3, 1], [-1, 2], [0,2]])
kmeans = KMeans(n_clusters=5, init = "random", n_init = 10, random_state=42)
y_pred = kmeans.fit_predict(X)

# Create a DataFrame for easier plotting with Plotly
df_clusters = pd.DataFrame(X, columns=['Feature 1', 'Feature 2'])
df_clusters['Cluster'] = y_pred.astype(str)

fig = px.scatter(df_clusters, x='Feature 1', y='Feature 2', color='Cluster',
                 title='K-Means Clustering Results (Plotly)',
                 labels={'Feature 1': 'Feature 1', 'Feature 2': 'Feature 2'})

# Add cluster centers to the plot
cluster_centers_df = pd.DataFrame(kmeans.cluster_centers_, columns=['Feature 1', 'Feature 2'])
fig.add_scatter(x=cluster_centers_df['Feature 1'], y=cluster_centers_df['Feature 2'],
                mode='markers', marker=dict(symbol='x', size=15, color='red', line=dict(width=2, color='black')),
                name='Cluster Centers')

fig.update_layout(showlegend=True)
fig.show()
```

```python
from sklearn.datasets import load_digits

X_digits, y_digits = load_digits(return_X_y=True)
X_train, y_train = X_digits[:1400], y_digits[:1400]
X_test, y_test = X_digits[1400:], y_digits[1400:]
```

```python
from sklearn.linear_model import LogisticRegression

n_labeled = 50
logistic_regression = LogisticRegression(max_iter=10000)
logistic_regression.fit(X_train[:n_labeled], y_train[:n_labeled])
logistic_regression.score(X_test, y_test)
```
